```python
import pandas as pd
from pathlib import Path

chemin_fichier = Path.cwd() / '../../data/raw/eco.csv'

# 1. Chargement avec tolérance des mauvaises lignes
# on_bad_lines='skip' ignorera silencieusement le dictionnaire de 16 colonnes à partir de la ligne 59
df_eco = pd.read_csv(
    chemin_fichier, 
    na_values='..', 
    encoding='cp1252',
    on_bad_lines='skip' 
)

# 2. Nettoyage de la pollution (Lignes 55 à 58)
# Les vraies données ont toujours un 'Country Code' (ex: BEN). 
# Les lignes de métadonnées auront un NaN ou du texte inutile ici.
# On supprime toutes les lignes où le code pays est manquant.
df_eco = df_eco.dropna(subset=['Country Code'])

# 3. Visualisation (Sanity Check)
print("--- 5 PREMIÈRES LIGNES ---")
display(df_eco.head())

print("\n--- 5 DERNIÈRES LIGNES (Pour vérifier la coupure) ---")
display(df_eco.tail())

print("\n--- STRUCTURE DU DATASET ---")
df_eco.info()
print(df_eco.columns)
```
